#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
tidy_data <- data |>
    # select the columns needed to answer research question
    select(tourney_finish, seed, conf_wins_pct, reg_wins_pct, first_game_at_home) |>
    # remove seed 0 since it was only used for 1 yr
    filter(seed != 0) |>
    # remove rows missing conference win percentage
    filter(!is.na(conf_wins_pct)) |>
    # make tourney_finish and first_game_at_home factors
    mutate(tourney_finish = factor(tourney_finish, 
                                    levels = c("first_round_loss", 
                                               "second_round_loss", "top_16_loss", 
                                               "top_8_loss", "top_4_loss", "top_2_loss", 
                                               "champ")), 
           first_game_at_home = as_factor(first_game_at_home))

# show first 6 rows
head(tidy_data)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
summary <- tidy_data |>
    # select the numerical columns to summarize
    select(seed, conf_wins_pct, reg_wins_pct) |>
    # put the variables all in 1 column so they can be grouped and summarized
    pivot_longer(c(seed, conf_wins_pct, reg_wins_pct), 
                 names_to = "variable", 
                 values_to = "value") |>
    # calculate the summary statistics
    group_by(variable) |>
    summarize(mean = mean(value, na.rm = TRUE), 
              min = min(value, na.rm = TRUE), 
              max = max(value, na.rm = TRUE))

summary
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
missing_table <- is.na(data) |>
    # convert it back into data frame to be summarized
    as.data.frame() |>
    # count the amount of true values in each column
    summarize(across(everything(), sum))
missing_table
```

```r
finish_summary <- tidy_data |>
    group_by(tourney_finish) |>
    summarize(mean_conf_wins_pct = mean(conf_wins_pct, na.rm = TRUE))

finish_summary
```

```r
conf_pct <- tidy_data |>
    ggplot(aes(x = conf_wins_pct)) + 
    geom_histogram(binwidth = 10) + 
    labs(x = "Conference Win Percentage", 
         y = "Number of Teams", 
         title = "Distribution of Conference Win Percentage") +
         theme(text = element_text(size = 15))

conf_pct
```

```r
finish_proportions <- tidy_data |>
    group_by(tourney_finish) |>
    summarize(n = n()) |>
    mutate(proportion = n / sum(n))

finish_graph <- finish_proportions |> 
    ggplot(aes(x = proportion, y = tourney_finish)) + 
    geom_bar(stat = "identity") +
    labs(x = "Proportion of Teams", 
         y = "Tournament Finish", 
         title = "Proportion of Teams Eliminated in Each Round") +
    theme(text = element_text(size = 15))

finish_graph
```
